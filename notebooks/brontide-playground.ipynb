## 3. Promote

A figure only counts once it is in `figures/`.


```python
def promote(fig, name, caption=""):
    """Write a figure to figures/ as the durable output.

    Anything not promoted is considered scratch and may vanish without loss.
    """
    out = FIGS / f"{name}.png"
    fig.savefig(out, dpi=200, bbox_inches="tight")
    print(f"wrote {out}" + (f"  # {caption}" if caption else ""))
    return out


# fig, ax = plt.subplots(figsize=(6, 4))
# ...
# promote(fig, "01-<what-it-shows>", "one sentence a reader needs")
```
